```python
goal_minutes = 30

exercise_minutes = int(input("วันนี้คุณออกกำลังกายกี่นาที: "))

if exercise_minutes >= goal_minutes:
    print("ยินดีด้วยครับ คุณออกกำลังกายถึงเป้าหมายแล้ว ")
else:
    remaining = goal_minutes - exercise_minutes
    print(f"ยังไม่ถึงเป้าหมาย ขาดอีก {remaining} นาที ")
```
